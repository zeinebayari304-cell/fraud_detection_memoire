```python
# ============================================================
# CELLULE 14 — ÉTAPE 13
# CONFIRMATION CLEAN SUR TEST
# AUCUN RÉGLAGE
# ============================================================

from sklearn.metrics import (

    accuracy_score,

    average_precision_score,

    confusion_matrix,

    f1_score,

    matthews_corrcoef,

    precision_score,

    recall_score,

    roc_auc_score
)


# ------------------------------------------------------------
# 1. Chargement Test
# ------------------------------------------------------------

X_test = pd.read_parquet(

    PROCESSED_DIR
    /
    "X_test.parquet"
)


y_test = (

    pd.read_parquet(

        PROCESSED_DIR
        /
        "y_test.parquet"
    )

    .squeeze()

    .astype(int)
)


if len(
    X_test
) != len(
    y_test
):

    raise RuntimeError(
        "X_test et y_test "
        "n'ont pas la même longueur."
    )


# ------------------------------------------------------------
# 2. Vérification espace de variables
# ------------------------------------------------------------

if "X_val" in globals():

    if list(
        X_test.columns
    ) != list(
        X_val.columns
    ):

        raise RuntimeError(
            "Les colonnes Test ne correspondent pas "
            "exactement aux colonnes Validation."
        )


# ------------------------------------------------------------
# 3. Fonction score continu
# ------------------------------------------------------------

def get_continuous_score(
    model,
    X
):

    # -----------------------------------------
    # Modèles avec probabilités
    # -----------------------------------------

    if hasattr(
        model,
        "predict_proba"
    ):

        probabilities = (
            model.predict_proba(
                X
            )
        )


        classes = getattr(
            model,
            "classes_",
            None
        )


        if (
            classes
            is not None
        ):

            classes_list = list(
                classes
            )


            if 1 not in (
                classes_list
            ):

                raise RuntimeError(
                    "Classe positive 1 "
                    "introuvable dans model.classes_."
                )


            positive_index = (

                classes_list.index(
                    1
                )
            )


            return (

                probabilities[
                    :,
                    positive_index
                ]
            )


        if (
            probabilities.ndim
            ==
            2

            and

            probabilities.shape[1]
            >=
            2
        ):

            return (
                probabilities[
                    :,
                    1
                ]
            )


    # -----------------------------------------
    # LinearSVC notamment
    # -----------------------------------------

    if hasattr(
        model,
        "decision_function"
    ):

        return (

            np.asarray(

                model.decision_function(
                    X
                )
            )

            .reshape(-1)
        )


    raise RuntimeError(
        "Impossible d'obtenir un score continu "
        "pour ce modèle."
    )


# ------------------------------------------------------------
# 4. Prédiction native
# ------------------------------------------------------------

y_test_pred = (

    np.asarray(

        final_model.predict(
            X_test
        )
    )

    .astype(int)
)


test_scores = (

    get_continuous_score(

        final_model,

        X_test
    )
)


# ------------------------------------------------------------
# 5. Matrice de confusion
# ------------------------------------------------------------

tn, fp, fn, tp = (

    confusion_matrix(

        y_test,

        y_test_pred,

        labels=[
            0,
            1
        ]
    )

    .ravel()
)


# ------------------------------------------------------------
# 6. Métriques
# ------------------------------------------------------------

final_test_df = pd.DataFrame(

    [
        {

            "Model":
                FINAL_MODEL_NAME,

            "Defense_Version":
                FINAL_MODEL_ROW[
                    "Defense_Version"
                ],

            "Accuracy":
                accuracy_score(
                    y_test,
                    y_test_pred
                ),

            "Precision":
                precision_score(
                    y_test,
                    y_test_pred,
                    zero_division=0
                ),

            "Recall":
                recall_score(
                    y_test,
                    y_test_pred,
                    zero_division=0
                ),

            "F1":
                f1_score(
                    y_test,
                    y_test_pred,
                    zero_division=0
                ),

            "MCC":
                matthews_corrcoef(
                    y_test,
                    y_test_pred
                ),

            "AP":
                average_precision_score(
                    y_test,
                    test_scores
                ),

            "ROC_AUC":
                roc_auc_score(
                    y_test,
                    test_scores
                ),

            "TN":
                int(
                    tn
                ),

            "FP":
                int(
                    fp
                ),

            "FN":
                int(
                    fn
                ),

            "TP":
                int(
                    tp
                )
        }
    ]
)


print(
    "=" * 95
)

print(
    "CONFIRMATION CLEAN — TEST"
)

print(
    "=" * 95
)


display(
    final_test_df.round(6)
)


# ------------------------------------------------------------
# 7. Export
# ------------------------------------------------------------

FINAL_TEST_FILE = (

    NB12_EXPORT_DIR

    /

    "11_final_model_test_clean.csv"
)


final_test_df.to_csv(

    FINAL_TEST_FILE,

    index=False
)


print(
    "\n✅ Test utilisé uniquement "
    "après le gel de la sélection NB12."
)

print(
    "✅ Aucun seuil, poids ou modèle "
    "n'est modifié à partir du Test."
)

print(
    "Export :",
    FINAL_TEST_FILE
)
```

```python
# ============================================================
# CELLULE 9 — ÉTAPE 8
# ANALYSE DE SENSIBILITÉ
# ============================================================

SCENARIOS = {

    # ---------------------------------------------
    # Règle principale
    # ---------------------------------------------

    "Équilibré": {

        "AP":
            0.25,

        "F1":
            0.25,

        "Recall":
            0.25,

        "HSJ_Robustness":
            0.25
    },


    # ---------------------------------------------
    # Accent sur la détection
    # ---------------------------------------------

    "Priorité détection": {

        "AP":
            0.20,

        "F1":
            0.25,

        "Recall":
            0.35,

        "HSJ_Robustness":
            0.20
    },


    # ---------------------------------------------
    # Accent sur la robustesse adversariale
    # ---------------------------------------------

    "Priorité robustesse": {

        "AP":
            0.15,

        "F1":
            0.15,

        "Recall":
            0.20,

        "HSJ_Robustness":
            0.50
    }
}


scenario_rows = []


for scenario_name, weights in (
    SCENARIOS.items()
):


    if not np.isclose(

        sum(
            weights.values()
        ),

        1.0
    ):

        raise RuntimeError(
            f"Poids invalides pour "
            f"'{scenario_name}'."
        )


    temp = (
        selection_df.copy()
    )


    temp[
        "Scenario_Score"
    ] = sum(

        weight
        *
        temp[
            criterion
        ]

        for criterion, weight
        in weights.items()
    )


    temp = (

        temp

        .sort_values(
            [
                "Scenario_Score",

                "AP",

                "MCC",

                "Model"
            ],

            ascending=[
                False,
                False,
                False,
                True
            ]
        )

        .reset_index(
            drop=True
        )
    )


    temp[
        "Rank"
    ] = (

        np.arange(
            1,
            len(temp)
            +
            1
        )
    )


    for _, row in (
        temp.iterrows()
    ):

        scenario_rows.append({

            "Scenario":
                scenario_name,

            "Model":
                row[
                    "Model"
                ],

            "Score":
                float(
                    row[
                        "Scenario_Score"
                    ]
                ),

            "Rank":
                int(
                    row[
                        "Rank"
                    ]
                )
        })


# ------------------------------------------------------------
# Tableau long
# ------------------------------------------------------------

sensitivity_df = (
    pd.DataFrame(
        scenario_rows
    )
)


expected_rows = (

    len(
        SCENARIOS
    )

    *

    len(
        selection_df
    )
)


if len(
    sensitivity_df
) != expected_rows:

    raise RuntimeError(
        "Le tableau de sensibilité est incomplet."
    )


# ------------------------------------------------------------
# Pivot des rangs
# ------------------------------------------------------------

rank_pivot = (

    sensitivity_df

    .pivot(
        index="Model",
        columns="Scenario",
        values="Rank"
    )
)


# ------------------------------------------------------------
# Gagnants
# ------------------------------------------------------------

scenario_winners = (

    sensitivity_df.loc[
        sensitivity_df[
            "Rank"
        ]
        ==
        1,

        [
            "Scenario",
            "Model",
            "Score"
        ]
    ]

    .sort_values(
        "Scenario"
    )

    .reset_index(
        drop=True
    )
)


print(
    "=" * 95
)

print(
    "ANALYSE DE SENSIBILITÉ — RANGS"
)

print(
    "=" * 95
)


display(
    rank_pivot
)


print(
    "\nGagnant de chaque scénario :"
)


display(
    scenario_winners.round(6)
)


# ------------------------------------------------------------
# Export
# ------------------------------------------------------------

SENSITIVITY_FILE = (

    NB12_EXPORT_DIR

    /

    "03_sensitivity_analysis.csv"
)


sensitivity_df.to_csv(
    SENSITIVITY_FILE,
    index=False
)


print(
    "\n✅ Analyse de sensibilité terminée."
)

print(
    "ℹ️ Les scénarios alternatifs mesurent "
    "la stabilité de la conclusion."
)

print(
    "ℹ️ Ils ne changent pas a posteriori "
    "la règle principale."
)
```

```python
# ============================================================
# CELLULE 8 — ÉTAPE 7
# SCORE MULTICRITÈRE PRINCIPAL
# ============================================================

PRIMARY_WEIGHTS = {

    "AP":
        0.25,

    "F1":
        0.25,

    "Recall":
        0.25,

    "HSJ_Robustness":
        0.25
}


# ------------------------------------------------------------
# Vérification de la somme des poids
# ------------------------------------------------------------

if not np.isclose(

    sum(
        PRIMARY_WEIGHTS.values()
    ),

    1.0
):

    raise RuntimeError(
        "La somme des poids doit être égale à 1."
    )


# ------------------------------------------------------------
# Calcul du score
# ------------------------------------------------------------

selection_df[
    "Primary_Score"
] = sum(

    weight
    *
    selection_df[
        criterion
    ]

    for criterion, weight
    in PRIMARY_WEIGHTS.items()
)


# ------------------------------------------------------------
# Classement
# ------------------------------------------------------------

primary_ranking = (

    selection_df

    .sort_values(
        [
            "Primary_Score",

            "AP",

            "MCC",

            "Model"
        ],

        ascending=[
            False,
            False,
            False,
            True
        ]
    )

    .reset_index(
        drop=True
    )
)


primary_ranking[
    "Primary_Rank"
] = (

    np.arange(
        1,
        len(
            primary_ranking
        )
        +
        1
    )
)


print(
    "=" * 95
)

print(
    "CLASSEMENT MULTICRITÈRE PRINCIPAL"
)

print(
    "25 % AP | 25 % F1 | "
    "25 % Recall | 25 % Robustesse HSJ"
)

print(
    "=" * 95
)


display(

    primary_ranking[
        [
            "Primary_Rank",

            "Model",

            "Defense_Version",

            "AP",

            "F1",

            "Recall",

            "HSJ_Robustness",

            "Primary_Score",

            "Pareto"
        ]
    ]

    .round(6)
)


# ------------------------------------------------------------
# Contrôle Pareto
# ------------------------------------------------------------

if not bool(
    primary_ranking.iloc[0][
        "Pareto"
    ]
):

    raise RuntimeError(
        "Le premier du score principal "
        "n'est pas Pareto-efficient."
    )


# ------------------------------------------------------------
# Export
# ------------------------------------------------------------

PRIMARY_RANKING_FILE = (

    NB12_EXPORT_DIR

    /

    "02_primary_ranking.csv"
)


primary_ranking.to_csv(
    PRIMARY_RANKING_FILE,
    index=False
)


print(
    "\n✅ Score principal calculé."
)

print(
    "✅ Aucune normalisation par rang utilisée."
)

print(
    "Export :",
    PRIMARY_RANKING_FILE
)
```

```python
# ============================================================
# CELLULE 7 — ÉTAPE 6
# FRONTIÈRE DE PARETO
# ============================================================

PARETO_CRITERIA = [

    "AP",

    "F1",

    "Recall",

    "HSJ_Robustness"
]


def is_dominated(
    row,
    dataframe,
    criteria,
    tolerance=1e-12
):

    """
    Un modèle est dominé s'il existe
    un autre modèle :

    - au moins aussi bon sur tous les critères ;
    - strictement meilleur sur au moins un critère.

    Tous les critères sont ici à maximiser.
    """

    for _, other in dataframe.iterrows():

        if (
            other[
                "Model"
            ]
            ==
            row[
                "Model"
            ]
        ):

            continue


        better_or_equal = all(

            other[c]
            >=
            row[c]
            -
            tolerance

            for c
            in criteria
        )


        strictly_better = any(

            other[c]
            >
            row[c]
            +
            tolerance

            for c
            in criteria
        )


        if (
            better_or_equal
            and
            strictly_better
        ):

            return True


    return False


# ------------------------------------------------------------
# Calcul Pareto
# ------------------------------------------------------------

selection_df[
    "Pareto"
] = (

    ~selection_df.apply(

        lambda row:

            is_dominated(
                row=row,
                dataframe=selection_df,
                criteria=PARETO_CRITERIA
            ),

        axis=1
    )
)


pareto_df = (

    selection_df[
        [
            "Model",

            "AP",

            "F1",

            "Recall",

            "HSJ_Robustness",

            "Pareto"
        ]
    ]

    .sort_values(
        [
            "Pareto",
            "AP",
            "F1"
        ],

        ascending=[
            False,
            False,
            False
        ]
    )

    .reset_index(
        drop=True
    )
)


print(
    "=" * 95
)

print(
    "FRONTIÈRE DE PARETO — VALIDATION"
)

print(
    "=" * 95
)


display(
    pareto_df.round(6)
)


pareto_models = (

    pareto_df.loc[
        pareto_df[
            "Pareto"
        ],
        "Model"
    ]

    .tolist()
)


if len(
    pareto_models
) == 0:

    raise RuntimeError(
        "Aucun modèle Pareto-efficient détecté."
    )


print(
    "\nModèles non dominés :",
    pareto_models
)


print(
    "✅ Analyse de Pareto réalisée "
    "avant toute pondération."
)
```

```python
# ============================================================
# ÉTAPE 4 — ROBUSTESSE HSJ DES VERSIONS FINALES
#            SUR VALIDATION
# VERSION CORRIGÉE ET ROBUSTE
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np


NB11_DIR = (
    PROJECT_ROOT
    / "notebooks"
    / "notebook11_adversarial_training"
)


print("=" * 90)
print("HSJ VALIDATION — VERSIONS FINALES DE NB11")
print("=" * 90)


# ============================================================
# 1. Fonction pour normaliser les noms des modèles
# ============================================================

def normalize_model_name(name):

    name = str(name).strip()

    aliases = {

        "Logistic Regression":
            "Régression logistique",

        "Regression logistique":
            "Régression logistique",

        "Régression Logistique":
            "Régression logistique",

        "Linear SVM":
            "SVM linéaire",

        "SVM":
            "SVM linéaire",

        "RandomForest":
            "Random Forest",

        "RF":
            "Random Forest",

        "XGB":
            "XGBoost",

        "LGBM":
            "LightGBM"
    }

    return aliases.get(
        name,
        name
    )


# ============================================================
# 2. Chercher tous les vrais fichiers HSJ Validation
# ============================================================

candidates = []


for path in sorted(
    NB11_DIR.rglob("*.csv")
):

    path_lower = str(path).lower()

    # On veut Validation
    if "validation" not in path_lower:
        continue

    # Pas les attaques servant à construire le Train adversarial
    if "adversarial_bank" in path_lower:
        continue

    if "hsj_train" in path_lower:
        continue

    try:

        df_tmp = pd.read_csv(path)

        if (
            "Model" in df_tmp.columns
            and
            "Local_Success" in df_tmp.columns
        ):

            df_tmp = df_tmp.copy()

            df_tmp["Model"] = (
                df_tmp["Model"]
                .apply(
                    normalize_model_name
                )
            )

            candidates.append(
                (
                    path,
                    df_tmp
                )
            )

    except Exception:
        pass


print(
    "\nNombre de fichiers candidats :",
    len(candidates)
)


for i, (path, df_tmp) in enumerate(
    candidates
):

    print("\n" + "-" * 90)

    print(
        f"[{i}]",
        path.relative_to(
            PROJECT_ROOT
        )
    )

    print(
        "Shape :",
        df_tmp.shape
    )

    print(
        "Modèles :",
        sorted(
            df_tmp[
                "Model"
            ]
            .dropna()
            .unique()
            .tolist()
        )
    )

    if "Condition" in df_tmp.columns:

        print(
            "Conditions :",
            df_tmp[
                "Condition"
            ]
            .astype(str)
            .unique()
            .tolist()
        )


# ============================================================
# 3. Identifier le fichier V2
#    LR + SVM
# ============================================================

v2_matches = [

    (
        path,
        df_tmp
    )

    for path, df_tmp
    in candidates

    if "defence_v2" in str(
        path
    ).lower()
]


if len(v2_matches) != 1:

    raise RuntimeError(
        f"Impossible d'identifier V2 : "
        f"{len(v2_matches)} fichier(s) trouvé(s)."
    )


V2_PATH, hsj_v2_raw = (
    v2_matches[0]
)


hsj_v2 = (
    hsj_v2_raw.copy()
)


# ------------------------------------------------------------
# Si V2 possède une colonne Condition :
# garder les lignes qui contiennent "defend".
#
# Si aucune ligne ne correspond, cela signifie que
# le fichier V2 est déjà un fichier final défendu :
# on garde alors toutes ses lignes.
# ------------------------------------------------------------

if "Condition" in hsj_v2.columns:

    condition_text = (
        hsj_v2[
            "Condition"
        ]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    defended_mask = (
        condition_text
        .str.contains(
            "defend",
            na=False
        )
    )

    if defended_mask.any():

        hsj_v2 = (
            hsj_v2[
                defended_mask
            ]
            .copy()
        )


print("\n" + "=" * 90)
print("FICHIER V2 RETENU")
print("=" * 90)

print(
    V2_PATH.relative_to(
        PROJECT_ROOT
    )
)

print(
    "Modèles V2 :",
    hsj_v2[
        "Model"
    ]
    .value_counts()
    .to_dict()
)


# ============================================================
# 4. Identifier le fichier V1
#    Il contient les cinq modèles + Original / Defended
# ============================================================

expected_five_models = {

    "Régression logistique",
    "SVM linéaire",
    "Random Forest",
    "XGBoost",
    "LightGBM"
}


v1_matches = []


for path, df_tmp in candidates:

    if (
        "defence_v2"
        in
        str(path).lower()
    ):
        continue

    if (
        "Condition"
        not in
        df_tmp.columns
    ):
        continue

    models_present = set(
        df_tmp[
            "Model"
        ]
        .dropna()
        .unique()
    )

    condition_text = (
        df_tmp[
            "Condition"
        ]
        .astype(str)
        .str.lower()
    )

    has_defended = (
        condition_text
        .str.contains(
            "defend",
            na=False
        )
        .any()
    )

    if (
        expected_five_models
        .issubset(
            models_present
        )
        and
        has_defended
    ):

        v1_matches.append(
            (
                path,
                df_tmp
            )
        )


if len(v1_matches) != 1:

    print(
        "\nCandidats V1 détectés :"
    )

    for path, df_tmp in v1_matches:

        print(
            "→",
            path.relative_to(
                PROJECT_ROOT
            )
        )

    raise RuntimeError(
        f"Impossible d'identifier V1 de manière unique : "
        f"{len(v1_matches)} fichier(s)."
    )


V1_PATH, hsj_v1_raw = (
    v1_matches[0]
)


# ============================================================
# 5. V1 : garder UNIQUEMENT la condition Defended
# ============================================================

condition_v1 = (
    hsj_v1_raw[
        "Condition"
    ]
    .astype(str)
    .str.strip()
    .str.lower()
)


hsj_v1_defended = (

    hsj_v1_raw[
        condition_v1
        .str.contains(
            "defend",
            na=False
        )
    ]

    .copy()
)


print("\n" + "=" * 90)
print("FICHIER V1 RETENU")
print("=" * 90)

print(
    V1_PATH.relative_to(
        PROJECT_ROOT
    )
)

print(
    "Modèles V1 défendus :",
    hsj_v1_defended[
        "Model"
    ]
    .value_counts()
    .to_dict()
)


# ============================================================
# 6. Fonction de résumé
# ============================================================

def summarize_hsj(
    dataframe
):

    result = (

        dataframe

        .groupby(
            "Model"
        )

        .agg(

            Attacks=(
                "Local_Success",
                "count"
            ),

            Successful_Attacks=(
                "Local_Success",
                "sum"
            )
        )

        .reset_index()
    )


    result[
        "HSJ_ASR"
    ] = (

        result[
            "Successful_Attacks"
        ]

        /

        result[
            "Attacks"
        ]
    )


    result[
        "HSJ_Robustness"
    ] = (

        1.0

        -

        result[
            "HSJ_ASR"
        ]
    )


    return result


v1_summary = summarize_hsj(
    hsj_v1_defended
)


v2_summary = summarize_hsj(
    hsj_v2
)


print("\n" + "=" * 90)
print("RÉSUMÉ V1 — DEFENDED")
print("=" * 90)

display(
    v1_summary
)


print("\n" + "=" * 90)
print("RÉSUMÉ V2 — FINAL")
print("=" * 90)

display(
    v2_summary
)


# ============================================================
# 7. Versions finales décidées dans NB11
# ============================================================

FINAL_DEFENSE_VERSION = {

    "Régression logistique":
        "V2",

    "SVM linéaire":
        "V2",

    "Random Forest":
        "V1",

    "XGBoost":
        "V1",

    "LightGBM":
        "V1"
}


# ============================================================
# 8. Assembler les CINQ versions finales
# ============================================================

final_hsj_rows = []


for model_name, version in (
    FINAL_DEFENSE_VERSION.items()
):

    if version == "V2":

        source = (
            v2_summary[
                v2_summary[
                    "Model"
                ]
                ==
                model_name
            ]
        )

    else:

        source = (
            v1_summary[
                v1_summary[
                    "Model"
                ]
                ==
                model_name
            ]
        )


    if len(source) == 0:

        print(
            "\n❌ Modèle non trouvé :",
            model_name,
            version
        )

        print(
            "\nModèles disponibles V1 :",
            v1_summary[
                "Model"
            ].tolist()
        )

        print(
            "Modèles disponibles V2 :",
            v2_summary[
                "Model"
            ].tolist()
        )

        raise RuntimeError(
            f"Résultat HSJ absent pour "
            f"{model_name} ({version})."
        )


    if len(source) > 1:

        raise RuntimeError(
            f"Plusieurs résultats trouvés pour "
            f"{model_name} ({version})."
        )


    row = (
        source.iloc[0]
    )


    final_hsj_rows.append({

        "Model":
            model_name,

        "Defense_Version":
            version,

        "Attacks":
            int(
                row[
                    "Attacks"
                ]
            ),

        "Successful_Attacks":
            int(
                row[
                    "Successful_Attacks"
                ]
            ),

        "HSJ_ASR":
            float(
                row[
                    "HSJ_ASR"
                ]
            ),

        "HSJ_Robustness":
            float(
                row[
                    "HSJ_Robustness"
                ]
            )
    })


hsj_summary = pd.DataFrame(
    final_hsj_rows
)


# ============================================================
# 9. Résultat FINAL
# ============================================================

print("\n" + "=" * 90)
print("HSJ VALIDATION — CINQ MODÈLES FINAUX DE NB11")
print("=" * 90)


display(
    hsj_summary[
        [
            "Model",
            "Defense_Version",
            "Attacks",
            "Successful_Attacks",
            "HSJ_ASR",
            "HSJ_Robustness"
        ]
    ]
    .sort_values(
        [
            "HSJ_ASR",
            "Model"
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# 10. CONTRÔLE STRICT
# ============================================================

assert len(
    hsj_summary
) == 5


assert set(
    hsj_summary[
        "Model"
    ]
) == expected_five_models


print("\n✅ Reconstruction HSJ réussie.")
print("✅ 5 modèles finaux présents.")
print("✅ LR et SVM → V2.")
print("✅ RF, XGBoost et LightGBM → V1.")
print("✅ Original et Defended ne sont pas mélangés.")
```

# Notebook 12 — Sélection multicritère du modèle final défendu

## 1. Objectif

Ce notebook constitue l'étape finale de sélection du classificateur après :

1. l'étude des modèles de référence ;
2. l'augmentation de la classe frauduleuse ;
3. l'évaluation des vulnérabilités adversariales ;
4. l'adversarial training.

Les modèles issus du Notebook 11 sont considérés comme définitivement entraînés.

**Aucun réentraînement, aucune modification des hyperparamètres,
aucune nouvelle augmentation et aucune optimisation du seuil de décision
ne sont réalisés dans ce notebook.**

Les modèles conservent leur règle de décision native.

---

## 2. Principe de sélection

La sélection finale doit tenir compte de deux dimensions complémentaires :

### Efficacité prédictive

Le système doit correctement distinguer les transactions frauduleuses des transactions normales.

Les métriques étudiées sont :

- Average Precision / PR-AUC ;
- Precision ;
- Recall ;
- F1-score ;
- MCC ;
- ROC-AUC.

En raison du fort déséquilibre des classes, l'Accuracy n'est pas utilisée
comme critère principal de sélection.

### Robustesse adversariale

La performance propre ne suffit pas pour sélectionner un système destiné
à être robuste.

La robustesse face à HSJ est donc intégrée à l'analyse à partir des résultats
obtenus sur Validation après adversarial training.

La mesure utilisée est :

Robustesse_HSJ = 1 - ASR_HSJ

où ASR représente le taux de succès des attaques locales admissibles.

---

## 3. Règle anti-fuite

La sélection du modèle final est effectuée exclusivement à partir des
résultats de Validation.

Les résultats Test ne participent pas au calcul du classement final.

Le Test est réservé à la confirmation finale du modèle sélectionné.

---

## 4. Pourquoi une analyse multicritère ?

Aucune métrique unique ne représente simultanément :

- la capacité à classer les fraudes rares ;
- la détection effective des fraudes ;
- le compromis Precision/Recall ;
- la robustesse face aux attaques adversariales.

La décision finale repose donc sur plusieurs critères complémentaires.

Afin d'éviter qu'un choix arbitraire de pondérations détermine seul le
résultat, trois niveaux d'analyse sont utilisés :

1. frontière de Pareto ;
2. score multicritère principal ;
3. analyse de sensibilité à plusieurs scénarios de pondération.

La sélection finale n'est donc pas fondée sur une métrique isolée.
